```python
import os
import numpy as np
import librosa
import librosa.display
import matplotlib.pyplot as plt
```

```python
def save_melspectrogram(audio_path, output_name):

    audio, sr = librosa.load(
        audio_path,
        sr=16000,
        mono=True
    )

    mel = librosa.feature.melspectrogram(
        y=audio,
        sr=sr,
        n_mels=128
    )

    mel_db = librosa.power_to_db(
        mel,
        ref=np.max
    )

    plt.figure(figsize=(8,4))

    librosa.display.specshow(
        mel_db,
        sr=sr,
        x_axis="time",
        y_axis="mel"
    )

    plt.colorbar(format="%+2.0f dB")

    plt.title(output_name)

    plt.tight_layout()

    plt.savefig(
        output_name + ".png",
        dpi=300
    )

    plt.show()

    plt.close()
```

```python
save_melspectrogram(
    "Dataset_Final_Indonesia/01_Anger/anger001.wav",
    "MelSpectrogram_Anger"
)

save_melspectrogram(
    "Dataset_Final_Indonesia/02_Sadness/sadness001.wav",
    "MelSpectrogram_Sadness"
)

save_melspectrogram(
    "Dataset_Final_Indonesia/03_Neutral/neutral001.wav",
    "MelSpectrogram_Neutral"
)

save_melspectrogram(
    "Dataset_Final_Indonesia/04_Happiness/happiness001.wav",
    "MelSpectrogram_Happiness"
)
```
